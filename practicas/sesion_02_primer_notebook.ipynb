```python
tiendas = 120
camaras_por_tienda = 3
lecturas_por_hora= 12
lecturas_al_dia = tiendas * camaras_por_tienda * lecturas_por_hora * 24
lecturas_al_anio = lecturas_al_dia * 365
print(f'Lecturas al día:{lecturas_al_dia}')
print(f'Lecturas al año: {lecturas_al_anio}')
```

```python
bytes_por_lectura = lecturas_al_dia * 46
megas_al_dia = bytes_por_lectura / 1000000
gigas_al_anio = (megas_al_dia *365)/1000
print(f'Sensores: {megas_al_dia:.2f} MB al día')
print(f'Sensores: {gigas_al_anio:.2f} GB al año')
```

```python
visitas = 250000
clics_por_visit = 12
clics_data_vol = 220
clics_al_dia = visitas * clics_por_visit
print(f'Clics al día: {clics_al_dia}')
megas_por_clics_al_dia = (clics_al_dia * clics_data_vol) /1000000
print(f'Clics: {megas_por_clics_al_dia:.0f} MB al día')
gigs_por_clics_al_ano = megas_por_clics_al_dia *365 / 1000
print(f'Clics: {gigs_por_clics_al_ano} GB al año')
```
