```python
print(pedido['cliente']['provincia'])
print(len(pedido['lineas']))
print(pedido['lineas'][1]['nombre'])
print(pedido['entrega']['tipo'])
print('descuento_pct' in pedido['lineas'][0])
pedido['revisado'] = True
total_price = 0
for line in pedido['lineas']:
  item_price = line['unidades'] * line['precio']
  if "descuento_pct" in line:
     item_price -= (item_price * line['descuento_pct'] /100)
  total_price += item_price

total_price += pedido['entrega']['coste_envio']
print(f'Total con envío: {total_price:.2f}€')
```

#¿por qué decimos que este dato es semiestructurado y no estructurado?
una fila, la columna correspondiente debe existir para todas las filas, independientemente de si las demás tienen un valor para esa columna o no.

Sin embargo, en los datos semiestructurados, esto funciona de manera diferente. No todos los registros tienen necesariamente los mismos campos, como ocurre con el campo descuento_pct en nuestro ejemplo.

| Dato | Tipo | Por qué |
|---|---|---|
| La tabla de tiendas de Colmena Market | Estructurado | Los datos siguen un esquema predefinido, con columnas fijas y tipos de datos consistentes. |
| Un pedido online en JSON | Semiestructurado | JSON tiene una estructura flexible basada en pares clave-valor, y no todos los registros contienen necesariamente los mismos campos. |
| La foto de un producto | No estructurado | El contenido visual de una imagen no sigue un esquema predefinido de campos y registros. |
| La reseña de un cliente | No estructurado | La reseña consiste en un texto libre que no sigue una estructura de datos predefinida. |
| La grabación de una llamada a atención al cliente | No estructurado | La grabación contiene datos de audio que no están organizados en campos o registros predefinidos. |
| El registro de accesos del servidor web | Estructurado / Semiestructurado | La clasificación depende de cómo se organizan los datos. Pueden seguir un esquema fijo o una estructura flexible con diferentes campos. |
| Una hoja de cálculo con los objetivos de ventas | Estructurado | Los datos están organizados en filas y columnas, con campos predefinidos y tipos de datos consistentes. |
| Un correo electrónico | Semiestructurado | Un correo electrónico contiene campos estructurados, como el remitente, el destinatario, el asunto y la fecha, pero su cuerpo puede contener texto no estructurado. |
